```python
# MERGE DATASETS
paths = {
    "2023_part01": "../data/contratos2023_part01.csv",
    "2023_part02": "../data/contratos2023_part02.csv",
    "2023_part03": "../data/contratos2023_part03.csv",
    "2024_part01": "../data/contratos2024_part01.csv",
    "2024_part02": "../data/contratos2024_part02.csv",
    "2024_part03": "../data/contratos2024_part03.csv",
    "2025_part01": "../data/contratos2025_part01.csv",
    "2025_part02": "../data/contratos2025_part02.csv",
    "2025_part03": "../data/contratos2025_part03.csv",
    "2026": "../data/contratos2026.csv",
}

datasets = {}

merged_dataset = pd.DataFrame()

for year, path in paths.items():
    print(f"Loading dataset for {year} from {path}...")
    datasets[year] = pd.read_csv(path)
    print(f"Dataset for {year} loaded successfully with shape {datasets[year].shape}.")
    merged_dataset = pd.concat([merged_dataset, datasets[year]], ignore_index=True)

print(f"Merged dataset created with shape {merged_dataset.shape}.")
```

## <font color='#BFD72F' size=6>3.4 Visualizations</font> <a class="anchor" id="3.4"></a>
  
[Back to TOC](#toc)

**Main Purpose**: understand which filters should we apply to have a smaller dataset (outlier detection, for example)

```python
contracts_by_month = (
    subset
    .dropna(subset=['dataCelebracaoContrato'])
    .assign(month=subset['dataCelebracaoContrato'].dt.to_period('M').dt.to_timestamp())
    .groupby('month')
    .size()
    .reset_index(name='number_of_contracts')
    .sort_values('month')
)

fig = px.line(
    contracts_by_month,
    x='month',
    y='number_of_contracts',
    title='Number of Contracts by Month',
    labels={
        'month': 'Month',
        'number_of_contracts': 'Number of Contracts'
    }
)

fig.show()
```

```python
# compute difference in days between publication and celebration, then plot histogram
diff_df = subset.dropna(subset=['dataPublicacao','dataCelebracaoContrato']).copy()
diff_df['days_diff'] = (diff_df['dataCelebracaoContrato'] - diff_df['dataPublicacao']).dt.days

fig_diff = px.histogram(
    diff_df,
    x='days_diff',
    nbins=100,
    title='Days between Publication and Celebration',
    labels={'days_diff':'Days difference', 'count':'Number of Contracts'}
)

fig_diff.update_xaxes(range=[int(diff_df['days_diff'].min()), int(diff_df['days_diff'].max())])
fig_diff.show()
```

```python
subset_counts = subset['tipoContrato'].value_counts().reset_index()
subset_counts.columns = ['tipoContrato', 'count']

# Get top 10
top_10 = subset_counts.head(10)

fig = px.bar(
    top_10,
    x='tipoContrato',
    y='count',
    title='Number of Contracts by Contract Type',
    labels={'tipoContrato': 'Contract Type', 'count': 'Number of Contracts'}
)

fig.show()
```

```python
subset_counts = subset['CPV'].value_counts().reset_index()
subset_counts.columns = ['CPV', 'count']

# Get top 10
top_10 = subset_counts.head(10)

fig = px.bar(
    top_10,
    x='CPV',
    y='count',
    title='Number of Contracts by CPV',
    labels={'CPV': 'CPV', 'count': 'Number of Contracts'}
)

fig.show()
```

```python
price_cols = ['precoBaseProcedimento', 'precoContratual', 'PrecoTotalEfetivo']

dropdown = widgets.Dropdown(
    options=price_cols,
    value=price_cols[0],
    description='Column:'
)

out = widgets.Output()

def plot_selected(col):
    data = subset[[col]].dropna().copy()
    fig = px.histogram(
        data,
        x=col,
        nbins=100,
        title=f'Distribution of {col}',
        labels={col: col}
    )
    fig.show()

def on_change(change):
    if change["name"] == "value":
        with out:
            clear_output(wait=True)
            plot_selected(change["new"])

dropdown.observe(on_change, names="value")

display(dropdown, out)

with out:
    plot_selected(dropdown.value)
```

```python
# Create a horizontal bar chart with LocalExecucao data
location_counts = subset['LocalExecucao'].value_counts().reset_index()
location_counts.columns = ['LocalExecucao', 'count']

# Get top 20 locations (sorted for better horizontal display)
top_20_locations = location_counts.head(20).sort_values('count', ascending=True)

fig = px.bar(
    top_20_locations,
    x='count',
    y='LocalExecucao',
    orientation='h',
    title='Number of Contracts by Execution Location (Top 20)',
    labels={'LocalExecucao': 'Location', 'count': 'Number of Contracts'}
)

fig.show()
```
