```python
print("There are {} Public Entities.".format(df['Entidade(s) Adjudicante(s)'].nunique()))
print("There are {} Companies.".format(df['Entidade(s) Adjudicatária(s)'].nunique()))

print("So, in total our analysis contains {} Nodes.".format(
    df['Entidade(s) Adjudicatária(s)'].nunique() + 
    df['Entidade(s) Adjudicante(s)'].nunique()))
```

```python
print("{} Contracts, characterised by {} columns.".format(df.shape[0], df.shape[1]))
```

```python
for col in df.columns:
    print("\n" + "="*60)
    print(f"Column: {col}")
    print("="*60)
    
    display(
        df[col]
        .value_counts(dropna=False)
        .reset_index()
        .rename(columns={"index": col, col: "count"})
    )
```

```python
cols_to_keep = ["Entidade(s) Adjudicante(s)",
                "Entidade(s) Adjudicatária(s)",
                "groupMembers",
                "Data de Celebração do Contrato",
                "Local de Execução",
                "CPV Designação",
                "Preço Total Efetivo"]
```

```python
df_subset = df[cols_to_keep]
df_subset.info() 
```

```python
for col in df_subset.columns:
    print("\n" + "="*60)
    print(f"Column: {col}")
    print("="*60)
    
    display(
        df[col]
        .value_counts(dropna=False)
        .reset_index()
        .rename(columns={"index": col, col: "count"})
    )
```

```python
df_counts = df['Tipo(s) de Contrato'].value_counts().reset_index()
df_counts.columns = ['Tipo(s) de Contrato', 'count']

bar_chart(
    data=df_counts,
    x='Tipo(s) de Contrato',
    y='count',
    title='Number of Contracts by Contract Type',
    labels={'Tipo(s) de Contrato': 'Contract Type', 'count': 'Number of Contracts'},
    top_n=20
)
```
