```python
df = df.drop(columns=[
    "Causa de Extinção do Contrato",
    "Data de Fecho do Contrato",
    "Preço Total Efetivo",
    "Causas das Alterações ao Prazo",
    "Causas das Alterações ao Preço",
    "Estado",
    "Aquisição ao Abrigo de Acordo Quadro de outro Estado-membro da UE",
    "Medidas especiais",
    "groupMembers"
])
```

```python
df["Preço Contratual"] = (
    df["Preço Contratual"]
    .str.replace("€", "")
    .str.replace(".", "")     
    .str.replace(",", ".")    
    .astype(float)
)
```

```python
df["Data de Publicação"] = pd.to_datetime(
    df["Data de Publicação"], 
    dayfirst=True, 
    errors="coerce"
)

df["Data de Celebração do Contrato"] = pd.to_datetime(
    df["Data de Celebração do Contrato"], 
    dayfirst=True, 
    errors="coerce"
)
```

```python
def detect_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    return series[(series < lower) | (series > upper)]
```

```python
price_outliers = detect_outliers_iqr(df["Preço Contratual"])
df.loc[price_outliers.index]
```

```python
import matplotlib.pyplot as plt

plt.boxplot(df["Preço Contratual"])
plt.title("Preço Contratual")
plt.show()
```

```python
import re

def parse_duration(x):
    if pd.isnull(x):
        return None
    x = x.lower()
    num = int(re.search(r'\d+', x).group())
    if "mes" in x:
        return num * 30
    elif "ano" in x:
        return num * 365
    else:
        return num

df["Prazo_dias"] = df["Prazo de Execução"].apply(parse_duration)
```

```python
plt.scatter(df["Prazo_dias"], df["Preço Contratual"])
plt.xlabel("Prazo (dias)")
plt.ylabel("Preço")
plt.show()
```
